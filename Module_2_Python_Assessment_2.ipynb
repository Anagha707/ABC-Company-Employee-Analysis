```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/drive/MyDrive/Entri Assignment/ABC Company Dataset.csv')

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
df.isnull().sum()
```

```python
np.random.seed(42)

df['Height'] = np.random.randint(150, 181, size=len(df))

df.head()
```

```python
print("Minimum Height:", df['Height'].min())

print("Maximum Height:", df['Height'].max())

print("Missing Height Values:", df['Height'].isnull().sum())
```

```python
df.to_csv('ABC_Company_Cleaned.csv', index=False)
```

Employee Distribution Across Each Team

```python
team_count = df['Team'].value_counts()

print("Employee Distribution Across Teams:")
print(team_count)
```

```python
#Calculating the percentage
team_percentage = (team_count / len(df)) * 100

print("Percentage Distribution Across Teams:")
print(team_percentage.round(2))
```

```python
#Combining the results into a table
team_analysis = pd.DataFrame({
    'Employee Count': team_count,
    'Percentage (%)': team_percentage.round(2)
})

team_analysis
```

```python
# Visualization
plt.figure(figsize=(15, 7))

sns.barplot(
    x=team_count.index,
    y=team_count.values
)

plt.title('Employee Distribution Across Teams')
plt.xlabel('Teams')
plt.ylabel('Number of Employees')
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

Segregate employees based on their positions

```python
#Count employees by position
position_count = df['Position'].value_counts()

print("Employee Distribution by Position:")
print(position_count)
```

```python
#Create a summary table
position_analysis = pd.DataFrame({
    'Employee Count': position_count
})

position_analysis
```

```python
#Bar Chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=position_count.index,
    y=position_count.values
)

plt.title('Employee Distribution by Position')
plt.xlabel('Position')
plt.ylabel('Number of Employees')

plt.tight_layout()
plt.show()
```

Identify the predominant age group

```python
#Create age groups
bins = [18, 25, 32, 40]

labels = ['19-25', '26-32', '33-40']

df['Age Group'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels
)
```

```python
#Count employees in each age group
age_group_count = df['Age Group'].value_counts().sort_index()

print("Employee Distribution by Age Group:")
print(age_group_count)
```

```python
#Identify predominant age group
predominant_age_group = age_group_count.idxmax()

print("Predominant Age Group:", predominant_age_group)

print("Number of Employees:", age_group_count.max())
```

```python
#Bar Chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=age_group_count.index,
    y=age_group_count.values
)

plt.title('Employee Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Employees')

plt.tight_layout()
plt.show()
```

Identify the highest salary expenditure

```python
#Salary expenditure by team
team_salary = df.groupby('Team')['Salary'].sum()

team_salary = team_salary.sort_values(ascending=False)

print("Salary Expenditure by Team:")
print(team_salary)
```

```python
# Identify the highest expenditure team
print("Team with Highest Salary Expenditure:")
print(team_salary.idxmax())

print("Total Salary Expenditure:")
print(team_salary.max())
```

```python
#Salary expenditure by position
position_salary = df.groupby('Position')['Salary'].sum()

position_salary = position_salary.sort_values(ascending=False)

print("Salary Expenditure by Position:")
print(position_salary)
```

```python
#Identify the highest expenditure position
print("Position with Highest Salary Expenditure:")
print(position_salary.idxmax())

print("Total Salary Expenditure:")
print(position_salary.max())
```

```python
#Team and position combination
team_position_salary = df.groupby(
    ['Team', 'Position']
)['Salary'].sum()

print("Top 10 Team and Position Salary Expenditure:")

print(
    team_position_salary.sort_values(
        ascending=False
    ).head(10)
)
```

```python
#Bar Chart (Team)
plt.figure(figsize=(15, 7))

team_salary.plot(kind='bar')

plt.title('Total Salary Expenditure by Team')
plt.xlabel('Team')
plt.ylabel('Total Salary')

plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

Correlation between age and salary

```python
correlation = df['Age'].corr(df['Salary'])

print("Correlation between Age and Salary:")
print(correlation)
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='Age',
    y='Salary'
)

plt.title('Correlation Between Age and Salary')
plt.xlabel('Age')
plt.ylabel('Salary')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    df[['Age', 'Salary']].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.3f'
)

plt.title('Age and Salary Correlation')

plt.tight_layout()
plt.show()
```
